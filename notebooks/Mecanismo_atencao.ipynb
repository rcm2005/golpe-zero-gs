# Laboratório de Atenção (Self-Attention no Braço)
Aqui vamos construir a fórmula fundamental do Transformer:
$$\text{Attention}(Q, K, V) = \text{softmax}\left(\frac{Q \cdot K^T}{\sqrt{d_k}}\right) \cdot V$$

Objetivo: Fazer cada palavra da frase calcular o quanto deve prestar atenção em cada uma das outras.

```python
import numpy as np
np.random.seed(42)

# 1. Frase de teste (3 palavras)
frase = ["banco", "não", "autorizou"]
N = len(frase) # 3 palavras
D = 4          # 4 dimensões no embedding

# 2. Embeddings das palavras da frase (Matriz X de tamanho 3 x 4)
# Cada linha é uma palavra, cada coluna é uma dimensão
X = np.random.randn(N, D) * 0.5

print("Frase:", frase)
print(f"Formato da matriz de entrada X (Palavras x Dimensões): {X.shape}\n")

# que merda é esse :<12 ali no meio da picaretagem, 
for palavra, vetor in zip(frase, X):
    print(f"{palavra:<12} -> {np.round(vetor, 3)}")
```

```python
# 3. As 3 Matrizes de Projeção do Transformer: W_q, W_k, W_v
# Elas pegam o embedding bruto (D=4) e transformam em Query, Key e Value (D=4)
W_q = np.random.randn(D, D) * 0.1  # Transforma em 'O que procuro?'
W_k = np.random.randn(D, D) * 0.1  # Transforma em 'O que ofereço?'
W_v = np.random.randn(D, D) * 0.1  # Transforma em 'Meu significado real'


```

```python

# Multiplica a frase (X) pelas 3 matrizes:
Q = np.dot(X, W_q)  # Formato (3, 4)
K = np.dot(X, W_k)  # Formato (3, 4)
V = np.dot(X, W_v)  # Formato (3, 4)



print("Setup pronto!")
print("Formato de Q:", Q.shape)
print("Formato de K:", K.shape)
print("Formato de V:", V.shape)
```
